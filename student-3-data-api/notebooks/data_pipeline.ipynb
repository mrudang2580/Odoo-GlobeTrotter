# 🌍 GlobeTrotter AI Data Pipeline & EDA Notebook

This notebook demonstrates:
1. Loading & profiling global city & activity datasets
2. Cost index normalization & correlation analysis
3. Similarity scoring & AI recommendation evaluation
4. Monte Carlo budget forecasting model

```python
import pandas as pd
import numpy as np

cities_df = pd.read_csv('../data/cities.csv')
activities_df = pd.read_csv('../data/activities.csv')

print('Cities Dataset Shape:', cities_df.shape)
print('Activities Dataset Shape:', activities_df.shape)
cities_df.head(3)
```

## 📊 1. Cost Distribution Across Regions

```python
region_summary = cities_df.groupby('region').agg({
    'cost_index': 'mean',
    'popularity_score': 'mean',
    'avg_daily_budget_usd': 'mean'
}).reset_index()

region_summary
```

## 🤖 2. Activity Categorization & Rating Insights

```python
category_stats = activities_df.groupby('category').agg({
    'cost': ['mean', 'median', 'max'],
    'rating': 'mean',
    'duration_minutes': 'mean'
})
category_stats
```
